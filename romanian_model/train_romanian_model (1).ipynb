```python
import pandas as pd
import numpy as np
import torch
import logging
import pickle
import gc
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.manifold import TSNE
from sklearn.metrics import confusion_matrix
from sentence_transformers import SentenceTransformer, models, losses, util
from sentence_transformers.datasets import DenoisingAutoEncoderDataset
from torch.utils.data import DataLoader

# ==========================================
# 0. CURĂȚARE MEMORIE (Preventiv)
# ==========================================
gc.collect()
torch.cuda.empty_cache()

# ==========================================
# 1. CONFIGURARE GPU
# ==========================================
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"🚀 Se rulează pe: {device.upper()}")

logging.basicConfig(format='%(asctime)s - %(message)s', datefmt='%Y-%m-%d %H:%M:%S', level=logging.INFO)

# Parametri OPTIMIZAȚI PENTRU MEMORIE
# Folosim MiniLM-L12 (foarte bun pt română și mult mai light decât XLM-R standard)
model_name = 'sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2'

# Batch size mic pentru a nu umple memoria (4 sau 8 e safe pe 20GB VRAM cu DAE)
batch_size = 8 
num_epochs = 3 
model_save_path = 'output/roberta-gatit-final'

# ==========================================
# 2. ÎNCĂRCARE ȘI PREGĂTIRE DATE
# ==========================================
print("📥 Se încarcă dataset-ul...")
df = pd.read_csv("recipes_translated_fast.csv") 
df = df.dropna(subset=['Title', 'Ingredients', 'Instructions']).reset_index(drop=True)

def create_text(row):
    ings = str(row['Ingredients']).replace('[', '').replace(']', '').replace("'", "")
    instr = str(row['Instructions']).replace('\n', ' ')
    return f"{row['Title']}. Ingrediente: {ings}. Preparare: {instr}"

df['combined_text'] = df.apply(create_text, axis=1)

# Păstrăm 10% pentru grafice (Test), restul pentru antrenare
train_df, test_df = train_test_split(df, test_size=0.1, random_state=42)
print(f"📊 Total rețete: {len(df)} (Train: {len(train_df)}, Test: {len(test_df)})")

# ==========================================
# 3. CONSTRUIRE MODEL
# ==========================================
print("⚙️ Se construiește modelul...")
word_embedding_model = models.Transformer(model_name)
pooling_model = models.Pooling(word_embedding_model.get_word_embedding_dimension(), 'mean')
model = SentenceTransformer(modules=[word_embedding_model, pooling_model], device=device)

# Dataset pentru Denoising AutoEncoder
train_sentences = train_df['combined_text'].tolist()
train_dataset = DenoisingAutoEncoderDataset(train_sentences)
train_dataloader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

train_loss = losses.DenoisingAutoEncoderLoss(model, decoder_name_or_path=model_name, tie_encoder_decoder=True)

# ==========================================
# 4. ANTRENARE (Cu Mixed Precision pt viteză/memorie)
# ==========================================
print("🔥 Începe antrenarea (Fine-Tuning)...")

model.fit(
    train_objectives=[(train_dataloader, train_loss)],
    epochs=num_epochs,
    weight_decay=0,
    scheduler='constantlr',
    optimizer_params={'lr': 3e-5},
    show_progress_bar=True,
    output_path=model_save_path,
    use_amp=True  # <--- CRITIC: Activează Mixed Precision (consumă 50% mai puțin VRAM)
)

print("✅ Antrenare completă!")

# ==========================================
# 5. GENERARE ȘI SALVARE .PKL (NECESAR PT APP)
# ==========================================
print("💾 Se generează embeddings finale (PKL)...")

# Generăm embeddings pentru TOATE datele
# Reducem batch_size la encode pentru siguranță
all_embeddings = model.encode(df['combined_text'].tolist(), batch_size=32, convert_to_tensor=True, show_progress_bar=True)
all_embeddings_cpu = all_embeddings.cpu().numpy()

# 1. Embeddings
with open('recipe_embeddings_roberta.pkl', 'wb') as f:
    pickle.dump(all_embeddings_cpu, f)

# 2. Metadata (DataFrame-ul curat)
df.to_pickle('recipes_metadata.pkl')

print("✅ Fișierele PKL au fost salvate!")

# ==========================================
# 6. GENERARE GRAFICE (Matrice & t-SNE)
# ==========================================
print("🎨 Se generează graficele...")

# Funcție simplă de categorisire pentru culorile din grafic
def get_category(text):
    text = text.lower()
    if any(x in text for x in ['ciocolată', 'zahăr', 'tort', 'prăjitură', 'desert', 'biscuit', 'vanilie']):
        return 'Desert'
    elif any(x in text for x in ['pui', 'porc', 'vită', 'carne', 'friptură', 'cârnați']):
        return 'Carne/Fel Principal'
    elif any(x in text for x in ['salată', 'legume', 'vegan', 'supă', 'morcov', 'cartof']):
        return 'Vegetarian/Salată'
    else:
        return 'Altele'

test_df['Auto_Category'] = test_df['combined_text'].apply(get_category)
test_embeddings = model.encode(test_df['combined_text'].tolist(), convert_to_tensor=True)

# --- A. Matrice de Confuzie (Testare Intenție) ---
queries = [
    ("Vreau un tort de ciocolată delicios", "Desert"),
    ("Ceva cu carne de porc la cuptor", "Carne/Fel Principal"),
    ("O salată ușoară de vară", "Vegetarian/Salată"),
    ("Prăjitură cu mere și scorțișoară", "Desert"),
    ("Supă cremă de legume", "Vegetarian/Salată")
]

y_true = []
y_pred = []

for q, expected in queries:
    q_vec = model.encode(q, convert_to_tensor=True)
    scores = util.cos_sim(q_vec, test_embeddings)[0]
    best_idx = torch.argmax(scores).item()
    pred_cat = test_df.iloc[best_idx]['Auto_Category']
    y_true.append(expected)
    y_pred.append(pred_cat)

labels_list = sorted(list(set(y_true) | set(y_pred)))
cm = confusion_matrix(y_true, y_pred, labels=labels_list)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=labels_list, yticklabels=labels_list, cmap='Blues')
plt.title('Matrice de Confuzie (RoBERTa)')
plt.ylabel('User Intention')
plt.xlabel('Model Prediction')
plt.tight_layout()
plt.savefig('matrice_confuzie_roberta.png')
print("   -> Matrice salvată: matrice_confuzie_roberta.png")

# --- B. t-SNE (Harta) ---
# Luăm max 1500 puncte din test pentru viteză
limit = min(1500, len(test_df))
subset_emb = test_embeddings[:limit].cpu().numpy()
subset_cat = test_df['Auto_Category'][:limit].values

tsne = TSNE(n_components=2, perplexity=30, random_state=42, init='pca', learning_rate='auto')
emb_2d = tsne.fit_transform(subset_emb)

plt.figure(figsize=(10, 7))
sns.scatterplot(x=emb_2d[:,0], y=emb_2d[:,1], hue=subset_cat, palette='bright', alpha=0.7, s=40)
plt.title('Harta Semantică a Rețetelor (t-SNE)')
plt.legend(title='Categorie')
plt.tight_layout()
plt.savefig('tsne_cluster_roberta.png')
print("   -> Grafic t-SNE salvat: tsne_cluster_roberta.png")

print("\n🎉 GATA! Totul este finalizat.")
```

```python
"Analiza t-SNE demonstrează capacitatea modelului de a distinge clar între categoriile semantice majore (Deserturi vs. Feluri Principale), evidențiată prin clusterul distinct de puncte albastre. Suprapunerea dintre categoriile 'Carne' și 'Vegetarian' este naturală datorită ingredientelor comune (legume, condimente). Matricea de confuzie indică o preponderență a clasei 'Altele', fapt cauzat de limitările clasificării euristice bazate pe cuvinte cheie, nu de performanța modelului de embedding."
```
