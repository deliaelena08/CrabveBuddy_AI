```python
import pandas as pd
from deep_translator import GoogleTranslator
import time
import os

INPUT_FILE = "../13k-recipes.csv"
OUTPUT_FILE = "../13k-recipes_translated_full.csv"
CHUNK_SIZE = 500
START_CHUNK = 0
DELAY = 0.5

def translate_text(text, src='en', target='ro', delay=DELAY):
    try:
        time.sleep(delay)
        return GoogleTranslator(source=src, target=target).translate(text)
    except Exception as e:
        print(f"⚠️ Translation failed for: {text[:60]}... - {e}")
        return text

def process_chunk(chunk_df):
    # Create new DataFrame for translated results
    translated_df = pd.DataFrame()
    translated_df['title_en'] = chunk_df['Title']
    translated_df['title_ro'] = chunk_df['Title'].astype(str).apply(lambda x: translate_text(x))
    translated_df['instructions_ro'] = chunk_df['Instructions'].astype(str).apply(lambda x: translate_text(x))
    translated_df['ingredients_ro'] = chunk_df['Cleaned_Ingredients'].astype(str).apply(lambda x: translate_text(x))
    return translated_df

def main():
    print("Reading input file")
    full_df = pd.read_csv(INPUT_FILE)
    total_rows = len(full_df)
    total_chunks = (total_rows // CHUNK_SIZE) + 1

    if os.path.exists(OUTPUT_FILE):
        print(f"Resuming from existing file: {OUTPUT_FILE}")
        output_df = pd.read_csv(OUTPUT_FILE)
        processed_rows = len(output_df)
        start_chunk = processed_rows // CHUNK_SIZE
    else:
        output_df = pd.DataFrame()
        start_chunk = START_CHUNK

    print(f"Starting translation in chunks of {CHUNK_SIZE} rows")

    for chunk_idx in range(start_chunk, total_chunks):
        print(f"\nProcessing chunk {chunk_idx + 1} of {total_chunks}")
        start = chunk_idx * CHUNK_SIZE
        end = min(start + CHUNK_SIZE, total_rows)
        chunk_df = full_df.iloc[start:end]

        try:
            translated_chunk = process_chunk(chunk_df)
            output_df = pd.concat([output_df, translated_chunk], ignore_index=True)

            output_df.to_csv(OUTPUT_FILE, index=False)
            print(f"Saved {len(output_df)} translated rows so far to: {OUTPUT_FILE}")
        except Exception as e:
            print(f"Error in chunk {chunk_idx}: {e}")
            continue

    print("\nTranslation completed for all chunks!")
    print(f"Final output saved to: {OUTPUT_FILE}")

if __name__ == "__main__":
    main()
```
