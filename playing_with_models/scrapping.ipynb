```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

options = Options()
# options.add_argument("--headless")
driver = webdriver.Chrome(options=options)

driver.get("https://jamilacuisine.ro/retete-video/dulciuri/")
wait = WebDriverWait(driver, 15)

try:
    wait.until(EC.frame_to_be_available_and_switch_to_it((By.CSS_SELECTOR, "iframe[src*='consent']")))
    accept_btn = wait.until(EC.element_to_be_clickable((By.XPATH, "//button[contains(., 'Accept')]")))
    accept_btn.click()
    print("Accept GDPR.")
    driver.switch_to.default_content()
    time.sleep(2)
except Exception as e:
    print("Nu a fost gasit iframe GDPR", e)

for _ in range(5):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)

links = set()
elems = driver.find_elements(By.CSS_SELECTOR, "a.penci-image-holder")
for e in elems:
    href = e.get_attribute("href")
    if href and "https://jamilacuisine.ro/" in href:
        links.add(href)

driver.quit()

print(f"Gasit {len(links)} linkuri.")
for l in links:
    print(" →", l)
```
